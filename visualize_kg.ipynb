```python
import os
import pickle as pkl

kg_file = "../knowledge_graphs_.pkl"
with open(kg_file, "rb") as f:
    kg = pkl.load(f)
```

```python
print(kg[0]["edge_indices"])
```

```python
import networkx as nx

# TODO add labels nodes and edges from openie subject/object/relations
def create_graph_from_edge_indices(edge_indices):
    edges = zip(*edge_indices)
    graph = nx.Graph()
    graph.add_edges_from(edges)
    return graph

edge_indices = kg[1]["edge_indices"].numpy().tolist()
graph = create_graph_from_edge_indices(edge_indices)
```

```python
import matplotlib.pyplot as plt

def visualize_graph(graph):
    nx.draw(graph, with_labels=True)
    plt.xlim(-0.5, 1.5)  # Set x-axis limits for zoom TODO fix this zoom
    plt.ylim(-0.5, 0.5)  # Set y-axis limits for zoom
    plt.show()
visualize_graph(graph)
```
